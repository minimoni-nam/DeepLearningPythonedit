```python
import pickle

filepath = 'citylearn_learning.pkl'

with (open(filepath, 'rb')) as f:
        model = pickle.load(f)

for n, nd in model.env.evaluate().groupby('name'):
    nd = nd.pivot(index='name', columns='cost_function', values='value').round(3)
    print(n, ':', nd.to_dict('records'))
```
